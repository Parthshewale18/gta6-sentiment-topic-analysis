#### Analysis & Visualize

```python
sentiment_map = {"positive": 1, "neutral": 0, "negative": -1}
df3["sentiment_numeric"] = df3["sentiment_label"].str.lower().map(sentiment_map)
df3.head()
```

```python
summary = df3[df3["topic_id"] != -1].groupby("topic_label").agg(
                      avg_sentiment=("sentiment_numeric", "mean"),
                      comment_count=("comment_id", "count"),).sort_values(
                      "comment_count", ascending=False).reset_index()
summary
```

```python
top = summary.head(10)
plt.figure(figsize=(10, 6))
colors = ["#d62728" if v < 0 else "#2ca02c" for v in top["avg_sentiment"]]
plt.barh(top["topic_label"], top["avg_sentiment"], color=colors)
plt.xlabel("Average sentiment (-1 = negative, +1 = positive)")
plt.title("GTA 6 Discussion: Sentiment by Topic")
plt.gca().invert_yaxis()
plt.tight_layout()
```

```python
sentiment_map = {"positive": 1, "neutral": 0, "negative": -1}
df3["sentiment_numeric"] = df3["sentiment_label"].str.lower().map(sentiment_map)
df3["date"] = pd.to_datetime(df3["date"])
daily = df3.groupby("date")["sentiment_numeric"].mean().reset_index()
plt.figure(figsize=(12, 5))
plt.plot(daily["date"], daily["sentiment_numeric"], marker="o", linewidth=1)
plt.axhline(0, color="gray", linestyle="--", linewidth=0.8)
plt.xlabel("Date")
plt.ylabel("Average sentiment")
plt.title("GTA 6 Sentiment Over Time")
#plt.xticks(rotation=45)
plt.tight_layout()
```

# This project set out to answer two questions about GTA 6 trailer comments: 
 what is the audience actually talking about, and how do
 they feel about each of those topics - not just an overall sentiment
 score, but sentiment broken down by theme and tracked over time.

# Sentiment by topic :
 Opinion is clearly NOT uniform - it splits sharply by topic, which a
 single overall sentiment score would have completely hidden:
- Most positive: pre-orders/preorder discussion (+0.37 avg
  sentiment) and trailer/pre-order chatter (+0.23) - the audience
  is genuinely excited about being able to commit to the game.
- Most negative: comments referencing "Rockstar Games" directly in
  a complaining context, and older "years ago" anticipation/
  impatience comments - these read as frustration with wait times
  and communication, not the game itself.
- Near-neutral: core topic discussion (graphics, "GTA VI and Grand")
  sits close to 0, suggesting descriptive/factual comments rather
  than strongly emotional ones.

# Sentiment over time :
 Daily average sentiment is highly volatile rather than flat or
 slowly drifting - it swings between roughly -0.3 and +0.55 day to
 day, with several sharp spikes (most notably a peak above +0.5 in
 early June). This pattern is consistent with sentiment being driven
 by specific triggering events (trailer drops, news, announcements)
 rather than a steady baseline mood, though pinning down the exact
 cause of each spike would need further investigation tying dates to
 news events.

# Overall takeaway:
 The audience holds two opinions at once - excitement about the game itself (setting, pre-orders) alongside frustration about the surrounding process (wait times, communication from Rockstar). This is the kind of nuance that combining unsupervised topic modeling with supervised sentiment analysis is able to surface, and that a single aggregate sentiment number cannot.
